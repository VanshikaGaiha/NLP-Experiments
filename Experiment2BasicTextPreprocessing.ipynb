### Experiment 2.1: Basic Text Preprocessing

Write a Python program to perform basic text preprocessing on given data. The program should:

1. **Convert text to lowercase** — Display the number of uppercase letters.
2. **Remove punctuation** — Display the number of different punctuation types.
3. **Remove numbers** — Display the number of numeric characters.
4. **Remove extra whitespace** — Display the amount of extra whitespace.
5. **Display the cleaned text.**

```python
import re
import string

# Read the dataset
with open("2.1_text_data.txt", "r") as file:
    text = file.read()

# 1. Count uppercase letters and convert to lowercase
uppercase = sum(1 for ch in text if ch.isupper())
text = text.lower()

# 2. Count different punctuation types and remove punctuation
punctuations = set(ch for ch in text if ch in string.punctuation)
text = text.translate(str.maketrans("", "", string.punctuation))

# 3. Count numbers and remove them
numbers = re.findall(r'\d+', text)
text = re.sub(r'\d+', '', text)

# 4. Count extra whitespace and remove it
extra_whitespace = len(text) - len(re.sub(r'\s+', ' ', text))
text = re.sub(r'\s+', ' ', text).strip()

# 5. Display results
print("Uppercase letters:", uppercase)
print("Different punctuation types:", len(punctuations))
print("Numbers:", len(numbers))
print("Extra whitespace:", extra_whitespace)

print("\nCleaned Text:")
print(text)
```

### Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
(Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

# Read the dataset
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

# Sentence tokenization
sentences = nltk.sent_tokenize(text)

# Word tokenization
words = nltk.word_tokenize(text)

print("SENTENCES:")
for sentence in sentences:
    print(sentence)

print("\nWORDS:")
print(words)
```

```python
import spacy

# Create a blank English pipeline
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")

# Read the dataset
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

doc = nlp(text)

# Sentence tokenization
print("SENTENCES:")
for sentence in doc.sents:
    print(sentence.text)

# Word tokenization
print("\nWORDS:")
for token in doc:
    print(token.text)
```

```python
import re

# Read the dataset
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()

# Sentence tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

# Word tokenization
words = re.findall(r'\w+|[^\w\s]', text)

print("SENTENCES:")
for sentence in sentences:
    print(sentence)

print("\nWORDS:")
print(words)
```

###Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens.
(Also, display Stop Words Separately).

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

# Read the dataset
with open("2.3_clean_data.txt", "r") as file:
    text = file.read()

# Tokenize the text
tokens = nltk.word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words('english'))

# Separate stop words and filtered tokens
found_stop_words = [word for word in tokens if word.lower() in stop_words]
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print("ORIGINAL TOKENS:")
print(tokens)

print("\nSTOP WORDS:")
print(found_stop_words)

print("\nFILTERED TOKENS:")
print(filtered_tokens)
```
