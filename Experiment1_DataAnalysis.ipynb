```python
# Importing all necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("employee_information_100.csv")
```

```python
# View the first 5 rows
print("First 5 Rows")
display(df.head())

# View column names, data types, and non-null counts
print("\nDataset Info")
df.info()
```

###Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.


```python
df.groupby('Department')['Salary'].mean().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Salary by Department')
plt.ylabel('Average Salary')
plt.xlabel('Department')
plt.tight_layout()
plt.show()
```

###Experiment 1.2: Display the number of employees working in each department.

```python
dept_counts = df['Department'].value_counts()
print("Number of employees in each department:\n")
print(dept_counts)
```

###Experiment 1.3:Display the percentage of male and female employees using a pie chart

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff'])
plt.title('Gender Distribution')
plt.show()
```

###Experiment 1.4: Draw a histogram showing salary distribution.

```python
df['Salary'].plot(kind='hist', bins=15, color='lightgreen', edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

###Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Experience_Years',
    y='Salary',
    hue='Department',
    style='Department',
    palette='tab10',
    s=80,
    linewidth=0.5,
)

plt.title('Experience vs. Salary by Department', fontsize=12, fontweight='bold')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Department', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

###Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10_earners = df.nlargest(10, 'Salary')
print("Top 10 Employees with Highest Salary:\n")
print(top_10_earners)
```

###Experiment 1.7: Find the highest salary in every department.

```python
max_salary_dept = df.groupby('Department')['Salary'].max()
print("Highest Salary in Each Department:\n")
print(max_salary_dept)
```

###Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_salary = df['Salary'].mean()
above_avg_employees = df[df['Salary'] > avg_salary]

print(f"Overall Average Salary: {avg_salary:.2f}\n")
print("Employees earning more than the average:\n")
print(above_avg_employees)
```

###Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_exp_dept = df.groupby('Department')['Experience_Years'].mean()
print('Average Years of Experience by Department:\n')
print(avg_exp_dept)
```

###Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
df['Age'].plot(kind='hist', bins=10, color='coral', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
